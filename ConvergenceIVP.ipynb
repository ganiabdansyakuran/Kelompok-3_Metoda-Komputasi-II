# Hasil Konvergensi untuk Masalah Nilai Awal (IVP)

Konvergensi untuk IVP sedikit berbeda dengan BVP, secara umum kita menginginkan
$$
    \lim_{\Delta t \rightarrow 0} U^N = u(t_f)
$$
di mana $t_f$ adalah waktu akhir yang diinginkan dan $N$ adalah jumlah langkah waktu yang diperlukan untuk mencapai $t_f$ sehingga
$$
    N \Delta t = t_f \quad \Rightarrow N = \frac{t_f}{\Delta t}.
$$

Namun, kita perlu berhati-hati pada tahap ini ketika kita berbicara tentang metode yang konvergen. Suatu metode dapat konvergen untuk satu set persamaan dan kondisi awal tertentu, tetapi tidak untuk yang lain. Secara praktis kita ingin hasil konvergensi berlaku untuk set persamaan dan kondisi awal yang cukup besar. Dengan pertimbangan tersebut, kita memiliki definisi berikut tentang konvergensi untuk IVP.

Jika menerapkan metode $r$-langkah pada PDB dengan bentuk
$$
    u'(t) = f(t,u)
$$
dengan $f(t,u)$ kontinu Lipschitz terhadap $u$, dan dengan set nilai awal apa pun yang memenuhi
$$
    \lim_{\Delta t\rightarrow 0} U^\nu(\Delta t) = u_0 \quad \text{untuk} \quad \nu = 0, 1, \ldots, r-1
$$
(yaitu penyalaan bootstrap untuk metode multi-langkah konsisten dengan nilai awal saat $\Delta t \rightarrow 0$), maka metode tersebut dikatakan *konvergen* dalam arti
$$
    \lim_{\Delta t \rightarrow 0} U^N = u(t_f).
$$

Seperti yang telah kita lihat sebelumnya, agar suatu metode menjadi konvergen, metode tersebut harus
 - **konsisten** - galat pemotongan lokal $\tau = \mathcal{O}(\Delta t^p)$ di mana $p > 0$ dan
 - **stabil-nol (zero-stable)** - bentuk stabilitas minimal serupa yang menyiratkan bahwa jumlah total galat saat $\Delta t \rightarrow 0$ terbatas dan memiliki orde yang sama dengan $\tau$ yang kita tahu menuju ke nol saat $\Delta t \rightarrow 0$.

## Konvergensi Metode Satu Langkah

Pertimbangkan masalah linier sederhana
$$
    \frac{\text{d}u}{\text{d}t} = \lambda u + g(t) \quad \text{dengan}\quad u(0) = u_0
$$
yang mana kita tahu memiliki solusi 
$$
    u(t) = u_0 e^{\lambda (t - t_0)} + \int^t_{t_0} e^{\lambda (t - \tau)} g(\tau) d\tau.
$$

### Euler Maju pada Masalah Linier

Menerapkan metode Euler pada masalah ini menghasilkan 
$$\begin{aligned}
    U^{n+1} &= U^n + \Delta t\lambda U^n \\
            &= (1 + \Delta t \lambda) U^n
\end{aligned}$$

Kita juga tahu galat pemotongan lokalnya adalah
$$\begin{aligned}
    \tau^n &= \left (\frac{u(t_{n+1}) - u(t_n)}{\Delta t} \right ) - \lambda u(t_n)\\
    &= \left (u'(t_n) + \frac{1}{2} \Delta t u''(t_n) + \mathcal{O}(\Delta t^2) \right ) - u'(t_n) \\
    &= \frac{1}{2} \Delta t u''(t_n) + \mathcal{O}(\Delta t^2)
\end{aligned}$$

Memperhatikan definisi asli $\tau^n$, kita dapat menulis ulang ekspresi untuk galat pemotongan lokal sebagai

$$
    u(t_{n+1}) = (1 + \Delta t \lambda) u(t_n) + \Delta t \tau^n
$$

yang mana jika dikombinasikan dengan penerapan metode Euler menghasilkan ekspresi untuk galat global

$$\begin{aligned}
    E^{n+1} = U^{n+1} - u(t^{n+1}) &= (1 + \Delta t \lambda) U^n - (1 + \Delta t \lambda) u(t_n) - \Delta t \tau^n \\
    &= (1+\Delta t \lambda) E^n - \Delta t \tau^n \\
\end{aligned}$$

Mengekspansi ekspresi ini mundur ke waktu $n=0$ menghasilkan 

$$
    E^n = (1 + \Delta t \lambda) E^0 - \Delta t \sum^n_{i=1} (1 + \Delta t \lambda)^{n-i} \tau^{i - 1}.
$$

Sekarang kita dapat melihat pentingnya suku $(1 + \Delta t \lambda)$. Kita dapat membatasi suku ini dengan

$$
    |1 + \Delta t \lambda| \leq e^{\Delta t \lambda}
$$

yang kemudian menyiratkan suku dalam penjumlahan dapat dibatasi oleh

$$
    |1 + \Delta t \lambda|^{n - i} \leq e^{(n-i) \Delta t |\lambda|} \leq e^{n \Delta t |\lambda||} \leq e^{|\lambda| t_f}
$$

Menggunakan ekspresi ini pada ekspresi galat global, kita temukan

$$\begin{aligned}
    E^n &= (1 + \Delta t \lambda) E^0 - \Delta t \sum^n_{i=1} (1 + \Delta t \lambda)^{n-i} \tau^{i - 1} \\
    |E^n| &\leq e^{|\lambda| \Delta t} |E^0| - \Delta t \sum^n_{i=1} e^{|\lambda| t_f} |\tau^{i - 1}| \\
          &\leq e^{|\lambda| t_f} \left(|E^0| - \Delta t \sum^n_{i=1} |\tau^{i - 1}|\right) \\
          &\leq e^{|\lambda| t_f} \left(|E^0| - n \Delta t \max_{1 \leq i \leq n} |\tau^{i - 1}|\right)
\end{aligned}$$

Dengan kata lain, galat global dibatasi oleh galat global awal dan galat satu-langkah maksimum yang dikalikan dengan jumlah langkah waktu yang diambil. Jika $N = \frac{t_f}{\Delta t}$ seperti sebelumnya dan dengan memperhitungkan galat pemotongan lokal, kita dapat menyederhanakan ekspresi ini lebih lanjut menjadi

$$
    |E^n| \leq e^{|\lambda| t_f} \left[|E^0| + t_f \left(\frac{1}{2} \Delta t |u''| + \mathcal{O}(\Delta t^2)\right ) \right]
$$

Jika kita asumsikan bahwa kita telah menggunakan kondisi awal yang benar $u_0$ maka $E_0 \rightarrow 0$ saat $\Delta t \rightarrow 0$ dan kita dapat melihat bahwa metode tersebut benar-benar konvergen karena 

$$
    |E^n| \leq e^{|\lambda| t_f} t_f \left(\frac{1}{2} \Delta t |u''| + \mathcal{O}(\Delta t^2)\right ) = \mathcal{O}(\Delta t).
$$

### Hubungan dengan Stabilitas untuk BVP

Kita dapat melihat hubungan antara versi stabilitas sebelumnya dengan yang diuraikan di atas. Cobalah menulis metode Euler maju sebagai sistem linier.

Euler Maju:
$$
    A = \frac{1}{\Delta t} \begin{bmatrix}
        1 \\
        -(1 + \Delta t \lambda) & 1 \\
        & -(1 + \Delta t \lambda) & 1 \\
        & & -(1 + \Delta t \lambda) & 1 \\
        & & & \ddots & \ddots \\
        & & & & -(1 + \Delta t \lambda) & 1 \\
        & & & & & -(1 + \Delta t \lambda) & 1
    \end{bmatrix}
$$
dengan
$$
    U = \begin{bmatrix} U^1 \\ U^2 \\ \vdots \\ U^N \end{bmatrix} ~~~~ 
    F = \begin{bmatrix} (1 / \Delta t + \lambda) U^0 + g(t_0) \\ g(t_1) \\ \vdots \\ g(t_{N-1}) \end{bmatrix}
$$

Mengikuti hasil stabilitas kita sebelumnya dan menetapkan $\hat{U~}$ sebagai vektor yang diperoleh dari solusi sejati ($\hat{U~}^i = u(t_i)$), maka kita punya
$$
    A U = F  ~~~~~~ A \hat{U~} = F + \tau
$$
dan oleh karena itu
$$
    A (\hat{U~} - U) = \tau.
$$

Mengingat bahwa $\hat{U~} - U = E$, kita kemudian dapat membalik matriks $A$ untuk mencari hubungan antara galat pemotongan $\tau$ dan galat global $E$. Seperti sebelumnya, kita mensyaratkan bahwa $A^{-1}$ dapat dibalik (yang trivial dalam kasus ini) dan bahwa $||A^{-1}|| < C$ dalam suatu norma tertentu. Kita dapat melihat hal ini sebagai
$$
    A^{-1} = \Delta t \begin{bmatrix}
        1 \\
        (1 + \Delta t \lambda) & 1 \\
        (1 + \Delta t \lambda)^2 & (1 + \Delta t \lambda) & 1 \\
        (1 + \Delta t \lambda)^3 & (1 + \Delta t \lambda)^2 & (1 + \Delta t \lambda) & 1\\
        \vdots & & & \ddots \\
        (1 + \Delta t \lambda)^{N-1} & (1 + \Delta t \lambda)^{N-2} (1 + \Delta t \lambda)^{N-3} & \cdots & (1 + \Delta t \lambda) & 1 
   \end{bmatrix}
$$
yang norma tak terhingganya adalah
$$
    ||A^{-1}||_\infty = \Delta t \sum^N_{m=1} | (1 + \Delta t \lambda)^{N-M} |
$$
dan oleh karena itu
$$
    ||A^{-1}||_\infty \leq \Delta t N e^{|\lambda| T} = T e^{|\lambda| T}.
$$
Saat $\Delta t \rightarrow 0$, ini terbatas untuk **T yang tetap**.

### Konvergensi Metode Satu Langkah Secara Umum

Pertimbangkan metode satu langkah umum yang dinotasikan dengan
$$
    U^{n+1} = U^n + \Delta t \Phi(U^n, t_n, \Delta t).
$$

Asumsikan $\Phi$ kontinu terhadap $t$ dan $\Delta t$ serta kontinu Lipschitz terhadap $u$ dengan konstanta Lipschitz $L$ (terkait dengan konstanta Lipschitz dari $f$). Jika metode satu langkah tersebut konsisten
$$
    \Phi(u,t,0) = f(u,t)
$$
untuk semua $u$, $t$, dan $\Delta t$ dan galat pemotongan lokalnya adalah
$$
    \tau^n =\frac{u(t_{n+1}) - u(t_n)}{\Delta t} - \Phi(u(t_n), t_n, \Delta t)
$$
maka metode satu langkah tersebut konvergen.  

Menggunakan pendekatan umum yang kita gunakan untuk Euler maju, kita tahu bahwa solusi sejati dan $\tau$ dihubungkan melalui
$$
    u(t_{n+1}) = u(t_n) + \Delta t \Phi(u(t_n), t_n, \Delta t) + \Delta t \tau^n
$$
yang mana jika dikurangkan dari solusi hampiran
$$
    U^{n+1} = U^n + \Delta t \Phi(U^n, t_n, \Delta t)
$$
menghasilkan
$$
    E^{n+1} = E^n + \Delta t (\Phi(U^n, t_n, \Delta t) - \Phi(u(t_n), t_n, \Delta t)) - \Delta t \tau^n.
$$

Menggunakan kekontinuan Lipschitz dari $\Phi$, maka kita punya
$$
    |E^{n+1}| \leq |E^n| + \Delta t L |E^n| + \Delta t |\tau^n|.
$$
yang memiliki bentuk yang sama seperti yang kita lihat pada pembuktian untuk Euler maju.

## Stabilitas-Nol untuk Metode Multi-langkah Linier

Kita juga dapat membuat pernyataan umum untuk metode multi-langkah linier, meskipun penting untuk dicatat bahwa kita memiliki persyaratan tambahan untuk metode multi-langkah linier agar konvergen. Sebagai contoh, pertimbangkan metode
$$
    U^{n+2} - 3 U^{n+1} + 2 U^n = - \Delta t f(U^n)
$$
sehingga $\alpha_0 = 2$, $\alpha_1 = -3$, dan $\alpha_2 = 1$ serta $\beta_0 = -1$ dengan sisanya sama dengan nol. Perhatikan bahwa koefisien-koefisien ini memenuhi syarat kita untuk konsisten dengan galat pemotongan
$$
    \tau^n = \frac{1}{\Delta t} (u(t_{n+2}) - 3 u(t_{n+1}) + 2 u(t_n) + \Delta t u'(t_n)) = \frac{5}{2} \Delta t u''(t_n) + \mathcal{O}(\Delta t^2).
$$
Ternyata meskipun metode ini konsisten, galat globalnya tidak konvergen secara umum!

$$
    U^{n+2} - 3 U^{n+1} + 2 U^n = - \Delta t f(U^n)
$$
Pertimbangkan metode di atas dengan PDB trivial
$$
    u'(t) = 0 \quad u(0) = 0
$$
sehingga kita mendapati metode tersebut menjadi
$$
    U^{n+2} - 3 U^{n+1} + 2 U^n = 0.
$$

Jika kita memiliki nilai eksak untuk $U^0$ dan $U^1$ maka metode ini akan menghasilkan $U^n = 0$. Namun secara umum kita hanya memiliki hampiran untuk $U^1$, lalu apa yang terjadi? Kita dapat memecahkan persamaan beda linier dalam kaitannya dengan $U^0$ dan $U^1$ untuk menemukan
$$
    U^n = 2 U^0 - U^1 + 2^n (U^1 - U^0).
$$
Jika kita mengasumsikan galat pada orde $\mathcal{O}(\Delta t)$ untuk $U^1$, ini dengan cepat mengarah ke nilai yang besar bahkan untuk $n$ yang kecil!

### Polinomial Karakteristik dan Persamaan Beda Linier

Sebagai selingan singkat, katakanlah kita ingin memecahkan

$$\sum^r_{j=0} \alpha_j U^{n+j} = 0$$

diberikan kondisi awal $U^0, U^1, \ldots, U^{r-1}$ yang memiliki solusi dalam bentuk umum $U^n = \xi^n$.  

Memasukkan ini ke dalam persamaan kita mendapatkan
$$
    \sum^r_{j=0} \alpha_j \xi^{n+j} = 0
$$
yang disederhanakan menjadi
$$
    \sum^r_{j=0} \alpha_j \xi^j = 0 
$$

dengan membaginya dengan $\xi^n$.  

Jika $\xi$ adalah akar dari polinomial
$$
    \rho(\xi) = \sum^r_{j=0} \alpha_j \xi^j
$$
maka $\xi$ memecahkan persamaan tersebut.

Perhatikan bahwa karena ini adalah metode linier maka kombinasi linier dari solusi-solusi juga merupakan solusi, sehingga bentuk umum solusinya berbentuk
$$
    U^n = c_1 \xi_1^n + c_2 \xi_2^n + \cdots + c_r \xi^n_r.
$$
Diberikan nilai awal untuk $U^0, U^1, \ldots$ kita dapat secara unik menentukan $c_j$.

### Hasil Stabilitas-Nol Umum untuk LMM

Suatu LMM $r$-langkah adalah *stabil-nol* jika akar-akar dari polinomial karakteristik $\rho(\xi)$ memenuhi
$$
    |\xi_j| \leq 1 \quad \quad \text{untuk} \quad j=1,2,3,\ldots,r
$$
jika $\xi_j$ bukan akar berulang dan $|\xi_j| < 1$ untuk akar berulang.

#### Contoh

Pertimbangkan metode multi-langkah linier
$$
    U^{n+2} - 2 U^{n+1} + U^n = \frac{\Delta t}{2} (f(U^{n+2}) - f(U^n)).
$$
Menerapkannya pada PDB $u'(t) = 0$ menghasilkan persamaan beda
$$
    U^{n+2} - 2 U^{n+1} + U^n = 0
$$
yang polinomial karakteristiknya adalah
$$
    \rho(\xi) = \xi^2 - 2 \xi + 1 = (\xi - 1)^2
$$
menghasilkan solusi umum
$$
    U^n = c_1 + c_2 n.
$$
Di sini kita melihat bahwa diberikan $U^0$ dan $U^1$ maka solusinya akan tetap tumbuh linier terhadap $n$ yang sekali lagi akan mengarah pada solusi yang divergen.

#### Contoh

Pertimbangkan metode multi-langkah linier
$$
    U^{n+3} - 2 U^{n+2} + \frac{5}{4} U^{n+1} - \frac{1}{4} U^n = \frac{\Delta t}{4} f(U^n).
$$
Terapkan ini pada PDB $u'(t) = 0$ dan tentukan apakah metode ini stabil-nol.

Diterapkan pada PDB $u'(t) = 0$ kita mendapatkan persamaan beda linier
$$
    U^{n+3} - 2 U^{n+2} + \frac{5}{4} U^{n+1} - \frac{1}{4} U^n = 0
$$
menghasilkan 
$$
    \rho(\xi) = \xi^3 - 2 \xi^2 + \frac{5}{4} \xi - \frac{1}{4} = 0
$$
yang solusi-solusinya adalah
$$
    \xi_1 = 1, \xi_2 = \xi_3 = 1 / 2
$$
dengan solusi umum
$$
    U^n = c_1 + c_2 \frac{1}{2^n} + c_3 n \frac{1}{2^n}
$$
yang memang konvergen karena adanya faktor $1 / 2^n$!

### Contoh: Metode Adams

Bentuk umum untuk semua metode Adams berbentuk
$$
    U^{n+r} = U^{n+r-1} + \Delta t \sum^r_{j=0} \beta_j f(U^{n+j})
$$
yang memiliki polinomial karakteristik (untuk PDB $u'(t)=0$)
$$
    \rho(\xi) = \xi^r - \xi^{r-1} = (\xi - 1) \xi^{r-1}
$$
menghasilkan akar $\xi_1 = 1$ dan $\xi_2 = \xi_3 = \cdots = \xi_r = 0$ yang memenuhi hasil stabilitas-nol secara umum dan oleh karena itu semua metode Adams adalah konvergen.

## Stabilitas Mutlak
Meskipun stabilitas-nol menjamin stabilitas, hal ini jauh lebih sulit digunakan secara umum karena batas $\Delta t \rightarrow 0$ dapat sulit dihitung. Sebagai gantinya, kita sering mempertimbangkan suatu $\Delta t$ yang terhingga dan memeriksa apakah metode tersebut stabil untuk pilihan $\Delta t$ tertentu ini. Keuntungan praktisnya adalah hal ini juga akan memberi tahu kita nilai $\Delta t$ tertentu mana yang akan memastikan bahwa metode kita memang stabil.

### Contoh
Pertimbangkan masalah

$$u'(t) = \lambda (u - \cos t) - \sin t \quad \text{dengan} \quad u(0) = 1$$

yang solusi eksaknya adalah 

$$u(t) = \cos t.$$

Kita dapat menghitung taksiran untuk nilai $\Delta t$ yang perlu kita gunakan dengan memeriksa galat pemotongan 

$$\begin{aligned}
    \tau &= \frac{1}{2} \Delta t u''(t) + \mathcal{O}(\Delta t^2) \\
      &= -\frac{1}{2} \Delta t \cos t + \mathcal{O}(\Delta t^2)
\end{aligned}$$

dan oleh karena itu

$$|E^n| \leq \Delta t \max_{0 \leq t \leq t_f} |\cos t| = \Delta t.$$

Jika kita menginginkan solusi dengan $|E^n| < 10^{-3}$ maka $\Delta t \approx 10^{-3}$. Beralih ke penerapan metode Euler, mari terapkan ini pada kasus di mana $\lambda = -10$ dan $\lambda = -2100$.

```python
# Compare accuracy between Euler
f = lambda t, lam, u: lam * (u - numpy.cos(t)) - numpy.sin(t)
u_exact = lambda t: numpy.cos(t)

t_f = 2.0
num_steps = [2**n for n in range(4, 10)]
# num_steps = [2**n for n in range(15,20)]
delta_t = numpy.empty(len(num_steps))
error_10 = numpy.empty(len(num_steps))
error_2100 = numpy.empty(len(num_steps))

for (i, N) in enumerate(num_steps):
    t = numpy.linspace(0, t_f, N)
    delta_t[i] = t[1] - t[0]
    
    # Compute Euler solution
    U = numpy.empty(t.shape)
    U[0] = 1.0
    for (n, t_n) in enumerate(t[1:]):
        U[n+1] = U[n] + delta_t[i] * f(t_n, -10.0, U[n])
    error_10[i] = numpy.abs(U[-1] - u_exact(t_f)) / numpy.abs(u_exact(t_f))
    
    U = numpy.empty(t.shape)
    U[0] = 1.0
    for (n, t_n) in enumerate(t[1:]):
        U[n+1] = U[n] + delta_t[i] * f(t_n, -2100.0, U[n])
    error_2100[i] = numpy.abs(U[-1] - u_exact(t_f)) / numpy.abs(u_exact(t_f))
    
# Plot error vs. delta_t
fig = plt.figure()
fig.set_figwidth(fig.get_figwidth() * 2)
axes = fig.add_subplot(1, 2, 1)

axes.loglog(delta_t, error_10, 'bo', label='Forward Euler')

order_C = lambda delta_x, error, order: numpy.exp(numpy.log(error) - order * numpy.log(delta_x))
axes.loglog(delta_t, order_C(delta_t[1], error_10[1], 1.0) * delta_t**1.0, 'r--', label="1st Order")

axes.legend(loc=4)
axes.set_title("Comparison of Errors")
axes.set_xlabel("$\Delta t$")
axes.set_ylabel("$|U(t_f) - u(t_f)|$")

axes = fig.add_subplot(1, 2, 2)

axes.loglog(delta_t, error_2100, 'bo', label='Forward Euler')
axes.loglog(delta_t, order_C(delta_t[1], error_2100[1], 1.0) * delta_t**1.0, 'r--', label="1st Order")

axes.set_title("Comparison of Errors")
axes.set_xlabel("$\Delta t$")
axes.set_ylabel("$|U(t_f) - u(t_f)|$")

plt.show()
```

Jadi apa yang salah dengan $\lambda = -2100$? Galat global seharusnya berjalan sebagai

$$E^{n+1} = (1 + \Delta t \lambda) E^n - \Delta t T^n$$

Jika $\Delta t \approx 10^{-3}$ maka untuk kasus $\lambda = -10$ galat global sebelumnya dikalikan dengan

$$1 + 10^{-3} \cdot -10 = 0.99$$

yang berarti kontribusi dari $E^n$ akan perlahan menurun seiring kita mengambil lebih banyak langkah waktu. Untuk kasus lainnya, kita punya

$$1 + 10^{-3} \cdot -2100 = -1.1$$

yang berarti bahwa untuk $\Delta t$ ini, galat yang dibuat pada langkah waktu sebelumnya akan bertambah besar! Agar hal ini tidak terjadi, kita harus membuat $\Delta t < 1 / 2100$ yang akan mengarah kembali pada konvergensi.

### Stabilitas Mutlak dari Metode Euler Maju

Pertimbangkan kembali masalah pengujian sederhana $u'(t) = \lambda u$. Kita tahu dari sebelumnya bahwa menerapkan metode Euler pada masalah ini menghasilkan pembaruan berbentuk

$$U_{n+1} = (1 + \Delta t \lambda) U_n.$$

Seperti yang mungkin sudah jelas dari contoh terakhir, kita tahu bahwa jika

$$|1 + \Delta t \lambda| \leq 1$$

maka metode tersebut akan stabil, ini disebut **stabilitas mutlak**. Perhatikan bahwa perkalian $\Delta t \lambda$ adalah hal yang penting di sini dan sering kali kita mempertimbangkan **daerah stabilitas mutlak** pada bidang kompleks yang didefinisikan oleh persamaan yang diuraikan di mana sekarang $z = \Delta t \lambda$. Ini memungkinkan nilai $\lambda$ menjadi kompleks yang bisa menjadi kasus penting untuk dipertimbangkan, terutama untuk sistem persamaan di mana $\lambda$ diidentifikasi sebagai nilai eigen.

```python
# Plot the region of absolute stability for Forward Euler
fig = plt.figure()
axes = fig.add_subplot(1, 1, 1)

t = numpy.linspace(0.0, 2.0 * numpy.pi, 100)

axes.fill(numpy.cos(t) - 1.0, numpy.sin(t), 'b')
axes.plot([-3, 3],[0.0, 0.0],'k--')
axes.plot([0.0, 0.0],[-3, 3],'k--')
axes.set_xlim((-3, 3.0))
axes.set_ylim((-3,3))
axes.set_aspect('equal')

axes.set_title("Absolute Stability Region for Forward Euler")
plt.show()
```

### Daerah Stabilitas Umum untuk Metode Multi-langkah Linier

Kembali ke metode multi-langkah linier dan menerapkannya secara umum pada masalah pengujian kita, kita mendapatkan
$$
    \sum^r_{j=0} \alpha_j U_{n+j} = \Delta t \sum^r_{j=0} \beta_j \lambda U_{n+j}
$$
yang dapat ditulis sebagai 
$$
    \sum^r_{j=0} (\alpha_j - \beta_j \Delta t \lambda) U_{n+j} = 0
$$
atau menggunakan notasi kita $z = \Delta t \lambda$ kita peroleh
$$
    \sum^r_{j=0} (\alpha_j - \beta_j z) U_{n+j} = 0.
$$

Ini memiliki bentuk yang mirip dengan persamaan beda linier yang dipertimbangkan di atas! Dengan menetapkan
$$
    \rho(\xi) = \sum^r_{j=0} \alpha_j \xi^j
$$
dan 
$$
    \sigma(\xi) = \sum^r_{j=0} \beta_j \xi^j
$$
kita dapat menulis ekspresi di atas sebagai
$$
    \pi(\xi, z) = \rho(\xi) - z \sigma(\xi)
$$
yang disebut **polinomial stabilitas** dari metode multi-langkah linier.  

Ternyata jika akar $\xi_i$ dari polinomial ini memenuhi
$$
    |\xi_i| \leq 1
$$
maka metode multi-langkah tersebut stabil-mutlak. Kita kemudian mendefinisikan daerah stabilitas mutlak sebagai nilai $z$ yang membuatnya bernilai benar. Pendekatan ini juga dapat diterapkan pada metode satu langkah.

### Contoh: Metode Euler Maju

Memeriksa metode Euler maju kita mendapatkan

$$\begin{aligned}
    0 &= U_{n+1} - U_n - \Delta t \lambda U_n \\
    &= U_{n+1} - U_n (1 + \Delta t \lambda)\\
    &= \xi - 1 (1 + z)\\
    &=\pi(\xi, z)
\end{aligned}$$

yang akarnya adalah $\xi = 1 + z$ dan kita telah menurunkan kembali daerah stabilitas yang kita temukan sebelumnya.

### Stabilitas Mutlak dari Metode Euler Mundur

Versi mundur dari metode Euler didefinisikan sebagai
$$
    U_{n+1} = U_n + \Delta t f(t_{n+1}, U_{n+1}).
$$
Periksa untuk melihat apakah Euler mundur stabil-mutlak.

Jika kita kembali mempertimbangkan masalah pengujian dari sebelumnya, kita temukan bahwa
$$\begin{aligned}
    0 &= U_{n+1} (1 - \Delta t \lambda) - U_n \\
    &= \xi (1 - z) - 1
\end{aligned}$$
yang memiliki akar $\xi = \frac{1}{1 - z}$. Kita kemudian mendapati
$$
    \left|\frac{1}{1-z}\right| \leq 1 \leftrightarrow |1 - z| \geq 1
$$
sehingga sebenarnya daerah stabilitasnya mencakup seluruh bidang kompleks kecuali lingkaran yang berpusat di $(1, 0)$ dengan jari-jari 1, yang menyiratkan bahwa metode Euler mundur sebenarnya stabil untuk pilihan $\Delta t$ mana pun.

## Penerapan pada PDB Kaku (Stiff ODEs)

Pertimbangkan kembali PDB yang kita periksa sebelumnya

$$u'(t) = \lambda (u - \cos t) - \sin t$$

kecuali kali ini dengan kondisi awal umum $u(t_0) = \eta$. Apa yang terjadi pada solusi yang sedikit berbeda dari $\eta = 1$ atau $t_0 = 0$? Solusi umum PDB tersebut adalah

$$u(t) = e^{\lambda (t - t_0)} (\eta - \cos t_0)) + \cos t$$.

```python
# Plot "hairy" solutions to the ODE
u = lambda t_0, eta, lam, t: numpy.exp(lam * (t - t_0)) * (eta - numpy.cos(t_0)) + numpy.cos(t)

fig = plt.figure()

for lam in [-1, -10]:
    fig = plt.figure()
    axes = fig.add_subplot(1, 1, 1)
    for eta in numpy.linspace(-1, 1, 10):
        for t_0 in numpy.linspace(0.0, 9.0, 10):
            t = numpy.linspace(t_0,10.0,100)
            axes.plot(t, u(t_0, eta, lam, t),'b')
    t = numpy.linspace(0.0,10.0,100)
    axes.plot(t, numpy.cos(t), 'r', linewidth=5)
        
    axes.set_title("Perturbed Solutions $\lambda = %s$" % lam)
    axes.set_xlabel('$t$')
    axes.set_ylabel('$u(t)$')
plt.show()
```

```python
# Plot "inverse hairy" solutions to the ODE
u = lambda t_0, eta, lam, t: numpy.exp(lam * (t - t_0)) * (eta - numpy.cos(t_0)) + numpy.cos(t)

fig = plt.figure()
num_steps = 10
error = numpy.ones(num_steps) * 1.0
t_hat = numpy.linspace(0.0, 10.0, num_steps + 1)
t_whole = numpy.linspace(0.0, 10.0, 1000)

fig = plt.figure()
axes = fig.add_subplot(1, 1, 1)
eta = 1.0
lam = 0.1

for n in range(1,num_steps):
    t = numpy.linspace(t_hat[n-1], t_hat[n], 100)
    U = u(t_hat[n-1], eta, lam, t)
    axes.plot(t, U, 'b')
    axes.plot(t_whole, u(t_hat[n-1], eta, lam, t_whole),'b--')
    axes.plot([t[-1], t[-1]], (U[-1], U[-1] + -1.0**n * error[n]), 'r')
    eta = U[-1] + -1.0**n * error[n]

t = numpy.linspace(0.0, 10.0, 100)
axes.plot(t, numpy.cos(t), 'g')

axes.set_title("Perturbed Solutions $\lambda = %s$" % lam)
axes.set_xlabel('$t$')
axes.set_ylabel('$u(t)$')
axes.set_ylim((-10,10))
plt.show()
```

### Contoh: Sistem Kimia

Pertimbangkan transisi bahan kimia $A$ ke bahan kimia $C$ melalui proses

$$A \overset{K_1}{\rightarrow} B \overset{K_2}{\rightarrow} C.$$

Jika kita menetapkan

$$\vec{u} = \begin{bmatrix} [A] \\ [B] \\ [C] \end{bmatrix}$$

maka kita dapat memodelkan reaksi kimia sederhana ini dengan sistem PDB

$$\frac{\text{d} \vec{u}}{\text{d} t} = 
\begin{bmatrix}
    -K_1 & 0 & 0 \\
    K_1 & -K_2 & 0 \\
    0 & K_2 & 0
\end{bmatrix} \vec{u}$$

Solusi dari sistem ini berbentuk

$$u_j(t) = c_{j1} e^{-K_1 t} + c_{j2}e^{-K_2 t} + c_{j3}$$

```python
# Solve the chemical systems example
# Problem parameters
K_1 = 3
K_2 = 1
# K_1 = 30.0
# K_2 = 1.0

A = numpy.array([[-K_1, 0, 0], [K_1, -K_2, 0], [0, K_2, 0]])
f = lambda u: numpy.dot(A, u)

t = numpy.linspace(0.0, 8.0, 128)
delta_t = t[1] - t[0]

U = numpy.empty((t.shape[0], 3))
U[0, :] = [2.5, 5.0, 2.0]

for n in range(t.shape[0] - 1):
    U[n+1, :] = U[n, :] + delta_t * f(U[n, :])
    
fig = plt.figure()
axes = fig.add_subplot(1, 1, 1)

axes.plot(t, U)

axes.set_title("Chemical System")
axes.set_xlabel("$t$")
axes.set_title("$[A], [B], [C]$")
axes.set_ylim((0.0, 10.))

plt.show()
```

### Apa itu Kekakuan (Stiffness)?

Secara umum suatu PDB yang **kaku (stiff)** adalah yang memenuhi $u'(t) \ll f'(t, u)$. Untuk sistem PDB, **rasio kekakuan (stiffness ratio)**

$$\frac{\max_p |\lambda_p|}{\min_p |\lambda_p|}$$

dapat digunakan untuk mengkarakterisasi kekakuan sistem. Pada contoh terakhir kita, rasio ini adalah $K_1 / K_2$ jika $K_1 > K_2$. Seiring kita meningkatkan rasio ini, kita mengamati bahwa metode numerik menjadi tidak stabil dan hanya pengurangan pada $\Delta t$ yang kembali menghasilkan solusi yang stabil. Untuk metode langkah waktu eksplisit, ini menjadi masalah karena pengurangan langkah waktu hanya untuk salah satu spesies mengarah pada perhitungan yang sangat mahal. Sebagai contoh, Euler maju memiliki kriteria stabilitas

$$|1 + \Delta t \lambda| < 1$$

di mana $\lambda$ harus berupa nilai eigen maksimum dari sistem tersebut.  

```python
# Plot the region of absolute stability for Forward Euler
fig = plt.figure()
axes = fig.add_subplot(1, 1, 1)

t = numpy.linspace(0.0, 2.0 * numpy.pi, 100)

K_1 = 3.0
K_2 = 1.0
delta_t = 1.0
eigenvalues = [-K_1, -K_2]

axes.fill(numpy.cos(t) - 1.0, numpy.sin(t), color='lightgray')
for lam in eigenvalues:
    print(lam * delta_t)
    axes.plot(lam * delta_t, 0.0, 'ko')
axes.plot([-3, 3],[0.0, 0.0],'k--')
axes.plot([0.0, 0.0],[-3, 3],'k--')
# axes.set_xlim((-3, 1))
axes.set_ylim((-2,2))
axes.set_aspect('equal')

axes.set_title("Absolute Stability Region for Forward Euler")
plt.show()
```

### Stabilitas-A
Bagaimana jika kita dapat memperluas daerah stabilitas mutlak untuk mencakup lebih banyak bagian bidang setengah kiri atau bahkan lebih baik, seluruhnya. Suatu metode yang memiliki sifat ini disebut **A-stabil**. Kita telah melihat salah satu contoh dari hal ini dengan Euler mundur yang memiliki daerah stabilitas

$$|1 - z| \geq 1$$

yang menutupi seluruh bidang setengah kiri.  

Ternyata untuk metode multi-langkah linier, suatu teorema oleh Dahlquist membuktikan bahwa tidak ada LMM yang memenuhi kriteria stabilitas-A melampaui orde kedua (aturan trapesium). Namun, ada metode Runge-Kutta orde lebih tinggi yang memenuhinya.

Mungkin ini terlalu ketat. Sering kali nilai eigen besar untuk sistem (misalnya yang berasal dari diskritisasi PDB parsial untuk persamaan panas) sepenuhnya terletak pada garis riil. Jika daerah stabilitas dapat mencakup sebanyak mungkin garis riil sambil menyisakan sisa bidang setengah kiri, kita bisa jadi mendapatkan metode yang lebih efisien. Ada sejumlah metode yang dapat dibangun yang memiliki sifat ini tetapi berorde lebih tinggi.

```python
# Plot the region of absolute stability for Backward Euler
fig = plt.figure()
axes = fig.add_subplot(1, 1, 1)

t = numpy.linspace(0.0, 2.0 * numpy.pi, 100)

K_1 = 3.0
K_2 = 1.0
delta_t = 1.0
eigenvalues = [-K_1, -K_2]

axes.set_facecolor('lightgray')
axes.fill(numpy.cos(t) + 1.0, numpy.sin(t), 'w')
for lam in eigenvalues:
    print(lam * delta_t)
    axes.plot(lam * delta_t, 0.0, 'ko')
axes.plot([-3, 3],[0.0, 0.0],'k--')
axes.plot([0.0, 0.0],[-3, 3],'k--')
# axes.set_xlim((-3, 1))
axes.set_ylim((-2,2))
axes.set_aspect('equal')

axes.set_title("Absolute Stability Region for Backward Euler")
plt.show()
```

### Stabilitas-L

Ternyata tidak semua metode A-stabil sama. Pertimbangkan metode Euler mundur dan metode trapesium. Polinomial stabilitas untuk metode trapesium adalah
$$\begin{aligned}
    0 &= U_{n+1} - U_n - \Delta t \frac{1}{2} (\lambda U_n + \lambda U_{n+1}) \\
      &= U_{n+1}\left(1 - \frac{1}{2} \Delta t \lambda \right ) - U_n \left(1 + \frac{1}{2}\Delta t \lambda \right) \\
      &= \left(\xi - \frac{1 + \frac{1}{2}z}{1 - \frac{1}{2} z}\right) \left(1 - \frac{1}{2} z \right )\\
\end{aligned}$$
yang menunjukkan bahwa ia A-stabil. Mari terapkan kedua metode ini pada masalah yang telah kita lihat sebelumnya dan lihat apa yang terjadi.

```python
# Compare accuracy between Euler
f = lambda t, lam, u: lam * (u - numpy.cos(t)) - numpy.sin(t)
u_exact = lambda t_0, eta, lam, t: numpy.exp(lam * (t - t_0)) * (eta - numpy.cos(t_0)) + numpy.cos(t)

t_0 = 0.0
eta = 1.5
lam = -1e6

num_steps = [10, 20, 40, 50]

delta_t = numpy.empty(len(num_steps))
error_euler = numpy.empty(len(num_steps))
error_trap = numpy.empty(len(num_steps))

for (i, N) in enumerate(num_steps):
    t = numpy.linspace(0, t_f, N)
    delta_t[i] = t[1] - t[0]
    u = u_exact(t_0, eta, lam, t_f)
    
    # Compute Euler solution
    U_euler = numpy.empty(t.shape)
    U_euler[0] = eta
    for (n, t_n) in enumerate(t[1:]):
        U_euler[n+1] = (U_euler[n] - lam * delta_t[i] * numpy.cos(t_n) - delta_t[i] * numpy.sin(t_n)) / (1.0 - lam * delta_t[i])
    error_euler[i] = numpy.abs(U_euler[-1] - u) / numpy.abs(u)
    
    # Compute using trapezoidal
    U_trap = numpy.empty(t.shape)
    U_trap[0] = eta
    for (n, t_n) in enumerate(t[1:]):
        U_trap[n+1] = (U_trap[n] + delta_t[i] * 0.5 * f(t_n, lam, U_trap[n]) - 0.5 * lam * delta_t[i] * numpy.cos(t_n) - 0.5 * delta_t[i] * numpy.sin(t_n)) / (1.0 - 0.5 * lam * delta_t[i])
    error_trap[i] = numpy.abs(U_trap[-1] - u)
    
# Plot error vs. delta_t
fig = plt.figure()
fig.set_figwidth(fig.get_figwidth() * 2)
axes = fig.add_subplot(1, 2, 1)
axes.plot(t, U_euler, 'ro-')
axes.plot(t, u_exact(t_0, eta, lam, t),'k')

axes = fig.add_subplot(1, 2, 2)
axes.loglog(delta_t, error_euler, 'bo')
order_C = lambda delta_x, error, order: numpy.exp(numpy.log(error) - order * numpy.log(delta_x))
axes.loglog(delta_t, order_C(delta_t[1], error_euler[1], 1.0) * delta_t**1.0, 'r--', label="1st Order")
axes.loglog(delta_t, order_C(delta_t[1], error_euler[1], 2.0) * delta_t**2.0, 'b--', label="2nd Order")

axes.legend(loc=4)
axes.set_title("Comparison of Errors for Backwards Euler")
axes.set_xlabel("$\Delta t$")
axes.set_ylabel("$|U(t_f) - u(t_f)|$")

# Plots for trapezoid
fig = plt.figure()
fig.set_figwidth(fig.get_figwidth() * 2)
axes = fig.add_subplot(1, 2, 1)
axes.plot(t, U_trap, 'ro-')
axes.plot(t, u_exact(t_0, eta, lam, t),'k')

axes = fig.add_subplot(1, 2, 2)
axes.loglog(delta_t, error_trap, 'bo', label='Forward Euler')
order_C = lambda delta_x, error, order: numpy.exp(numpy.log(error) - order * numpy.log(delta_x))
axes.loglog(delta_t, order_C(delta_t[1], error_trap[1], 1.0) * delta_t**1.0, 'r--', label="1st Order")
axes.loglog(delta_t, order_C(delta_t[1], error_trap[1], 2.0) * delta_t**2.0, 'b--', label="2nd Order")

axes.legend(loc=4)
axes.set_title("Comparison of Errors for Trapezoidal Rule")
axes.set_xlabel("$\Delta t$")
axes.set_ylabel("$|U(t_f) - u(t_f)|$")

plt.show()
```

Ternyata jika kita melihat pada metode satu langkah dan mendefinisikan rasio berikut

$$U_{n+1} = R(z) U_n$$

kita dapat mendefinisikan bentuk stabilitas lainnya, yang disebut **Stabilitas-L**, di mana kita mensyaratkan bahwa metode tersebut A-stabil dan bahwa

$$\lim_{z \rightarrow \infty} |R(z)| = 0.$$

Euler mundur adalah L-stabil sedangkan metode trapesium tidak.

## Rumus Beda Mundur (Backward Differencing Formulas / BDF)

Suatu kelas metode LMM yang berguna untuk masalah PDB kaku adalah metode rumus beda mundur (BDF) yang memiliki bentuk

$$\alpha_0 U_n + \alpha_1 U_{n+1} + \cdots + \alpha_r U_{n+r} = \Delta \beta_r f(U_{n+r})$$

Metode-metode ini dapat diturunkan langsung dari beda hingga mundur dari titik $U_{n+r}$ dan sisa titik mundur ke belakang dalam waktu. Seseorang kemudian dapat menurunkan metode r-langkah yang akurat orde ke-r dengan cara ini. Beberapa dari metodenya adalah 

$$\begin{aligned}
    r = 1:& & U_{n+1} - U_n = \Delta t f(U_{n+1}) \\
    r = 2:& &3 U_{n+2} - 4 U_{n+1} + U_n = 2 \Delta t f(U_{n+1}) \\
    r = 3:& &11U_{n+3} - 18U_{n+2} + 9U_{n+1} - 2 U_n = 6 \Delta t f(U_{n+3}) \\
    r = 4:& &25 U_{n+4} - 48 U_{n+3} +36 U_{n+2} -16 U_{n+1} +3 U_n = 12 \Delta t f(U_{n+4})
\end{aligned}$$

## Plotting Daerah Stabilitas

Jika kita menganggap akar dari polinomial stabilitas $\xi_j$ sebagai bilangan kompleks dan menulisnya dalam bentuk eksponensial 
$$
    \xi_j = |\xi_j| e^{i \theta}.
$$

Di sini $|\xi_j|$ adalah modulus (atau besaran) dari bilangan kompleks dan didefinisikan sebagai $|\xi_j| = x^2 + y^2$ di mana $\xi_j = x + i j$. Jika $\xi_j$ berada pada batas daerah stabilitas mutlak maka kita tahu bahwa $|\xi_j| = 1$. Menggunakan hal ini bersama dengan polinomial stabilitas kemudian menghasilkan 
$$
    \rho(e^{i\theta}) - z \sigma(e^{i\theta}) = 0
$$
yang mana dengan memecahkannya untuk $z$ menghasilkan
$$
    z(\theta) = \frac{\rho(e^{i\theta})}{\sigma(e^{i\theta})}.
$$

Sebagai contoh pertimbangkan metode 2-tahap Adams-Bashforth. Polinomial stabilitasnya dapat ditemukan sebagai
$$\begin{aligned}
    U_{n+2} &= U_{n+1} + \frac{\Delta t}{2} (-f(U_n) + 3 f(U_{n+1})) \\
    \pi(\xi, z) &= U_{n+2} - U_{n+1} - \frac{\Delta t}{2} (-f(U_n) + 3 f(U_{n+1})) = 0 \\
    &= U_{n+2} - U_{n+1} - \frac{1}{2} (\Delta t \lambda U_n - 3 \Delta t \lambda U_{n+1}) \\
    &= 2 \xi^2 - 2 \xi + 3 z\xi - z \\
    &= \rho(\xi, z) + z \sigma(\xi, z)
\end{aligned}$$
di mana
$$
    \rho(\xi, z) = 2 ( \xi - 1) \xi ~~~ \text{dan} ~~~ \sigma(\xi, z) = 3 \xi - 1
$$
sehingga
$$
    z(\theta) = \frac{2 (\xi - 1) \xi}{3 \xi - 1}.
$$

Hal ini tidak serta merta memastikan bahwa jika diberikan suatu $\theta$ maka $z(\theta)$ akan terletak pada batas daerah stabilitas mutlak. Ini dapat terjadi ketika $\xi_j = 1$ tetapi di sisi kiri dan kanan kurva $\xi_j > 1$ dan oleh karena itu ia tidak menandai batas daerah tersebut. Untuk menentukan apakah suatu daerah tertentu yang diuraikan oleh kurva ini berada di dalam atau di luar daerah stabilitas, kita dapat mengevaluasi semua akar dari $\pi(\xi, z)$ pada suatu $z$ di dalam daerah yang dipertanyakan dan melihat apakah kemudian $\forall j, \xi_j < 1$.

Untuk metode satu langkah, ini menjadi lebih mudah, jika kita melihat pada rasio $R(z)$ yang kita definisikan sebelumnya sebagai
$$
    U_{n+1} = R(z) U_n
$$
dalam kasus metode deret Taylor orde ke-p yang diterapkan pada $u'(t) = \lambda u$ kita mendapatkan

$$\begin{aligned}
    U_{n+1} &= U_n + \Delta t \lambda U_n + \frac{1}{2}\Delta t^2 \lambda^2 U_n + \cdots + \frac{1}{p!}\Delta t^p \lambda^p U_n \\
    &=\left(1 + z + \frac{1}{2} z^2 + \frac{1}{6} z^3 + \cdots +\frac{1}{p!}z^p\right) U_n \Rightarrow \\
    R(z) &= 1 + z + \frac{1}{2} z^2 + \frac{1}{6} z^3 + \cdots +\frac{1}{p!}z^p.
\end{aligned}$$

Menetapkan $R(z) = e^{i\theta}$ dapat mengarah pada cara untuk menyelesaikan pencarian batas (di mana $|R(z)| = 1$) tetapi hal ini sangat sulit dilakukan secara umum. Sebagai gantinya jika kita memplot kontur dari $R(z)$ di bidang kompleks, kita dapat memilih kontur $R(z)=1$ dan memplotnya.

```python
theta = numpy.linspace(0.0, 2.0 * numpy.pi, 100)

# ==================================
#  Forward euler
fig = plt.figure()
fig.set_figwidth(fig.get_figwidth() * 2.0)
fig.set_figheight(fig.get_figheight() * 2.0)
axes = fig.add_subplot(2, 2, 1)
x = numpy.linspace(-2, 2, 100)
y = numpy.linspace(-2, 2, 100)
X, Y = numpy.meshgrid(x, y)
Z = X + 1j * Y

# Evaluate which regions are actually in stability region
print("Forward Euler's method - Stability region tests")
z = -1 + 1j * 0
print("  Inside of circle: ", numpy.abs(1.0 + z))
z = -3 + 1j * 0
print("  Outside of circle: ", numpy.abs(1.0 + z))

axes.contour(X, Y, numpy.abs(1.0 + Z), levels=[1.0])
axes.plot(x, numpy.zeros(x.shape),'k')
axes.plot(numpy.zeros(y.shape), y,'k')
axes.set_aspect('equal')
axes.set_title("Forward Euler")

# ==================================
#  Backwards Euler
axes = fig.add_subplot(2, 2, 2)
x = numpy.linspace(-2, 2, 100)
y = numpy.linspace(-2, 2, 100)
X, Y = numpy.meshgrid(x, y)
Z = X + 1j * Y

# Evaluate which regions are actually in stability region
print("Backward Euler's method - Stability region tests")
z = 1 + 1j * 0
print("  Inside of circle: ", numpy.abs(1.0 + z))
z = -1 + 1j * 0
print("  Outside of circle: ", numpy.abs(1.0 + z))

axes.contour(X, Y, numpy.abs(1.0 / (1.0 - Z)), levels=[1.0])
axes.plot(x, numpy.zeros(x.shape),'k')
axes.plot(numpy.zeros(y.shape), y,'k')
axes.set_aspect('equal')
axes.set_title("Backwards Euler")

# ==================================
#  Taylor series method of order 4
axes = fig.add_subplot(2, 2, 3)
x = numpy.linspace(-5, 5, 100)
y = numpy.linspace(-5, 5, 100)
X, Y = numpy.meshgrid(x, y)
Z = X + 1j * Y

# Evaluate which regions are actually in stability region
print("Taylor series method of order 4 - Stability region tests")
z = -1 + 1j * 0
print("  Inside of strange region: ", numpy.abs(1.0 + z))
z = -4 + 1j * 0
print("  Outside of strange region: ", numpy.abs(1.0 + z))

axes.contour(X, Y, numpy.abs(1 + Z + 0.5 * Z**2 + 1.0/6.0 * Z**3 + 1.0 / 24.0 * Z**4), levels=[1.0])
axes.plot(x, numpy.zeros(x.shape),'k')
axes.plot(numpy.zeros(y.shape), y,'k')
axes.set_aspect('equal')
axes.set_title("4th Order Taylor Series")

# ==================================
# 2-step Adams-Bashforth
theta = numpy.linspace(0.0, 2.0 * numpy.pi, 1000)
xi = numpy.exp(1j * theta)

rho_2AB = lambda xi: (xi - 1.0) * xi
sigma_2AB = lambda xi: (3.0 * xi - 1.0) / 2.0
z_2AB = rho_2AB(xi) / sigma_2AB(xi)
z = rho_2AB(xi) / sigma_2AB(xi)

axes = fig.add_subplot(2, 2, 4)
axes.plot(z_2AB.real, z_2AB.imag)
axes.plot(x, numpy.zeros(x.shape),'k')
axes.plot(numpy.zeros(y.shape), y,'k')
axes.set_title("2-step Adams-Bashforth")
axes.set_aspect('equal')
axes.set_xlim([-2, 3])
axes.set_ylim([-2, 2])

plt.show()
```
