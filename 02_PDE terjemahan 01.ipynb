# Review: Persamaan Diferensial Parsial (PDE)

Pada materi ini kita akan meninjau kembali konsep dasar yang paling penting sebagai latar belakang untuk pembahasan metode numerik pada persamaan diferensial parsial (PDE). Sebagian besar yang akan digunakan adalah pengetahuan umum tentang klasifikasi PDE, teknik penyelesaian untuk tiap jenis PDE, serta beberapa sifat solusinya. Ini juga mencakup topik seperti well-posedness, kondisi batas, dan hal terkait lainnya.

Pertama, kita mulai dengan notasi. Kita akan menggunakan penulisan:
$$
\frac{\partial u}{\partial x} = u_x
$$
untuk menyatakan turunan parsial fungsi (u) terhadap variabel (x).

Turunan orde lebih tinggi atau turunan campuran ditulis seperti:
$$
\frac{\partial^2 u}{\partial x^2} = u_{xx}, \quad
\frac{\partial^2 u}{\partial y \partial x} = u_{xy}
$$
Perlu diingat bahwa secara umum urutan turunan bisa berpengaruh.

Kapan turunan bisa ditukar urutannya?
Di sini kita menggunakan variabel yang biasanya terkait koordinat Kartesius, tetapi kita juga bisa menggunakan koordinat lain seperti polar:
$$
\frac{\partial u}{\partial \theta} = u_\theta
$$
atau terhadap waktu:
$$
\frac{\partial u}{\partial t} = u_t.
$$

## 1. Klasifikasi PDE

Klasifikasi klasik PDE didasarkan pada bentuk umum PDE linear orde dua:
$$
a u_{xx} + 2b u_{xy} + c u_{yy} + d u_x + e u_y + f u = g.
$$

Perlu dicatat bahwa dalam klasifikasi umum, variabel (x) atau (y) bisa diganti dengan (t) atau variabel lainnya. Jika kita ambil transformasi Fourier dari persamaan ini, bentuknya menjadi:
$$
P(X,Y) = a X^2 + 2b X \cdot Y + c Y^2 + \cdots
$$
yang menyerupai polinomial orde dua.

Istilah klasifikasi PDE berasal dari fakta bahwa bentuk ini mirip dengan irisan kerucut (conic sections), yang biasanya diklasifikasikan menjadi eliptik, parabolik, atau hiperbolik berdasarkan diskriminannya:
$$
b^2 - ac.
$$

Maka klasifikasi PDE ditentukan dari tanda diskriminan:

* Jika $(b^2 - ac < 0)$ → PDE eliptik
* Jika $(b^2 - ac = 0)$ → PDE parabolik
* Jika $(b^2 - ac > 0)$ → PDE hiperbolik

Mengapa klasifikasi ini penting? Karena biasanya ada metode numerik tertentu yang cocok untuk satu jenis PDE, tetapi tidak cocok untuk jenis lain. Kita akan membahas lebih lanjut nanti, tetapi sekarang mari lihat contoh dasar dari tiap klasifikasi.

### Contoh

### **Persamaan Laplace**

Dalam dimensi (n), persamaan Laplace:
$$
\Delta u = \nabla^2 u = \sum^n_{i=1} u_{x_i x_i} = 0
$$

Untuk kasus 2 dimensi:
$$
u_{xx} + u_{yy} = 0.
$$

Jenis PDE apa ini?
Koefisiennya: (a = 1), (b = 0), (c = 1). Maka:
$$
b^2 - ac = -1.
$$
Karena negatif, persamaan Laplace adalah PDE **eliptik**.

### **Persamaan Panas (Heat Equation)**

Dalam satu dimensi ruang:
$$
u_t = u_{xx}.
$$

Jenis PDE apa ini?
Persamaan panas adalah **parabolik**. Jika turunan terhadap (y) diganti dengan (t), diperoleh:
(a = -1), (b = 0), (c = 0), sehingga diskriminan = 0.

Bagaimana jika lebih dari satu dimensi ruang?
$$
u_t = u_{xx} + u_{yy}.
$$
Klasifikasinya jadi lebih kompleks karena melibatkan tiga variabel.

### **Persamaan Gelombang (Wave Equation)**

Dalam satu dimensi:
$$
u_{tt} - u_{xx} = 0.
$$

Dari eliminasi, ini adalah PDE **hiperbolik**.
Dengan mengganti turunan (y) menjadi (t):
(a = -1), (b = 0), (c = 1)

$$
b^2 - ac = -(-1)(1) = 1.
$$

### **Persamaan Navier–Stokes**

Digunakan untuk memodelkan aliran fluida tak termampatkan:

[
\begin{aligned}
u_t + (u^2)*x + (uv)*y &= - \frac{1}{\rho} p_x + \nu (u*{xx} + u*{yy}) \
v_t + (uv)*x + (v^2)*y &= - \frac{1}{\rho} p_y + \nu (v*{xx} + v*{yy})
\end{aligned}
]


Persamaan ini sulit diklasifikasikan secara sederhana. Di dalamnya terdapat:

* Bagian mirip hiperbolik (sebelah kiri),
* Bagian eliptik/parabolik (sebelah kanan, terutama karena difusi).

Karena itu, Navier–Stokes sering dianggap **parabolik**, tetapi metode penyelesaiannya bisa mengambil teknik dari berbagai klasifikasi.

### Cara Pandang yang Lebih Umum

Cara lain yang lebih intuitif untuk memahami klasifikasi PDE adalah melihat bagaimana informasi merambat dalam domain:

* **PDE Hiperbolik:** Informasi merambat dengan kecepatan terbatas.
* **PDE Parabolik:** Informasi menyebar tak terbatas (contoh: difusi).
* **PDE Eliptik:** Kasus batas antara hiperbolik dan parabolik dalam perambatan informasi.
