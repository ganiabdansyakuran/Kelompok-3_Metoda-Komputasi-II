# Persamaan Eliptik

## Persamaan Eliptik

Persamaan diferensial parsial orde dua secara umum:

$$
a_1 u_{xx} + a_2 u_{xy} + a_3 u_{yy} + a_4 u_x + a_5 u_y + a_6 u = f
$$

dengan:
- $u_{xx}$ = turunan kedua terhadap $x$
- $u_{yy}$ = turunan kedua terhadap $y$
- $u_{xy}$ = turunan campuran
- $u_x, u_y$ = turunan pertama
- $a_1, a_2, ..., a_6$ = koefisien
- $f$ = fungsi sumber

### Syarat Persamaan Eliptik

$$
a_2^2 - 4 a_1 a_3 < 0
$$

Jika syarat ini terpenuhi, maka persamaan termasuk tipe eliptik.

### Penjelasan

Nilai diskriminan yang negatif menunjukkan bahwa persamaan tidak memiliki arah karakteristik nyata, sehingga solusi yang dihasilkan bersifat halus (smooth) dan tidak menunjukkan perilaku gelombang.

### Contoh

Persamaan Laplace:

$$
u_{xx} + u_{yy} = 0
$$

Cek:
$$
0^2 - 4(1)(1) = -4 < 0
$$

Maka termasuk persamaan eliptik.

Salah satu tempat alami di mana persamaan seperti ini muncul kembali adalah pada masalah keadaan tunak (steady-state) yang berasal dari persamaan panas.
Dalam dua dimensi kita punya
$$ u_t=(\kappa u_x)_x+(\kappa u_y)_y+\Psi (x,y). $$
dengan:
- $ u(x,y,t)$ = suhu sebagai fungsi ruang dan waktu,
- $ \kappa $  = konduktivitas termal,
- $ \Psi (x,y) $ = sumber panas.

Jika keadaan tunak berarti suhu tidak berubah terhadap waktu, sehingga:
 $ u_t=0$, maka kita memperoleh 
$$ (\kappa u_x)_x+(\kappa u_y)_y=-\Psi (x,y). $$



Jika kita mengasumsikan $\kappa$  konstan, maka kita dapat menuliskan ulang persamaan ini sebagai
$$ u_{xx}+u_{yy}=-\frac{\Psi }{\kappa }=f(x,y), $$
yang merupakan bentuk berdimensi lebih tinggi dari persamaan Poisson (dan persamaan Laplace jika $$ f(x,y)=0). $$



## Contoh Persamaan Eliptik (Penjelasan Sederhana)

Diberikan persamaan:

$$
2u_{xx} + 4u_{xy} + 3u_{yy} + u_x + 2u_y + u = x^2 + y^2
$$


## Langkah Penyelesaian

### 1. Ambil koefisien utama

Dari persamaan:

- $a_1 = 2$
- $a_2 = 4$
- $a_3 = 3$


### 2. Gunakan syarat eliptik

$$
a_2^2 - 4a_1 a_3 < 0
$$

Hitung:

$$
4^2 - 4(2)(3) = 16 - 24 = -8
$$


### 3. Kesimpulan

Karena hasilnya negatif, maka:

Persamaan ini adalah persamaan eliptik


## Penjelasan Singkat

- Persamaan eliptik biasanya digunakan untuk kondisi stabil (steady-state)
- Contohnya:
  - suhu
  - potensial listrik
  - medan gravitasi


## Intinya

Cukup cek:

$$
a_2^2 - 4a_1 a_3
$$

- < 0 → eliptik  
- = 0 → parabolik  
- > 0 → hiperbolik

## 5-Point Stencil

Ekstensi dari pendekatan satu dimensi ke dua dimensi memungkinkan kita membangun pendekatan beda hingga untuk operator Laplasian, yang dikenal sebagai 5-point stencil:

$$
\nabla^2 u \approx 
\frac{U_{i-1,j} - 2U_{i,j} + U_{i+1,j}}{\Delta x^2}
+
\frac{U_{i,j-1} - 2U_{i,j} + U_{i,j+1}}{\Delta y^2}
$$

### Penjelasan

Persamaan ini merupakan pendekatan numerik untuk operator Laplace $\nabla^2 u$ dalam dua dimensi menggunakan metode beda hingga (finite difference).

- $U_{i,j}$ = nilai fungsi pada titik grid $(i,j)$  
- $U_{i\pm1,j}$ = titik tetangga arah $x$  
- $U_{i,j\pm1}$ = titik tetangga arah $y$  
- $\Delta x, \Delta y$ = jarak antar grid  

Disebut 5-point stencil karena melibatkan 5 titik:
- 1 titik pusat $(i,j)$
- 4 titik di sekelilingnya (atas, bawah, kiri, kanan)

### Intuisi Fisik

Pendekatan ini menyatakan bahwa nilai Laplasian di suatu titik bergantung pada perbedaan nilai di titik tersebut terhadap tetangga sekitarnya. Dalam banyak kasus fisika (seperti panas dan potensial), ini menggambarkan proses difusi atau penyebaran.

### Kasus Khusus (Grid Seragam)

Jika $\Delta x = \Delta y = h$, maka persamaan menjadi lebih sederhana:

$$
\nabla^2 u \approx 
\frac{U_{i-1,j} + U_{i+1,j} + U_{i,j-1} + U_{i,j+1} - 4U_{i,j}}{h^2}
$$

Jika $$ \Delta x=\Delta y, $$ kita dapat menyederhanakan ekspresi ini menjadi
$$ \nabla ^2u\approx \frac{U_{i-1,j}+U_{i+1,j}-4U_{i,j}+U_{i,j-1}+U_{i,j+1}}{\Delta x^2}. $$



Kita dapat menghampiri solusi dari persamaan Poisson sebagai
$$ \frac{U_{i-1,j}+U_{i+1,j}-4U_{i,j}+U_{i,j-1}+U_{i,j+1}}{\Delta x^2}=f_{i,j}, $$
yang menghasilkan $ m^2 $ persamaan untuk $ m^2 $ variabel tak diketahui.


## Contoh Soal 5-Point Stencil

Diberikan pendekatan Laplacian dengan 5-point stencil:

$$
\nabla^2 u \approx 
\frac{U_{i-1,j} - 2U_{i,j} + U_{i+1,j}}{\Delta x^2}
+
\frac{U_{i,j-1} - 2U_{i,j} + U_{i,j+1}}{\Delta y^2}
$$


## Soal

Diketahui nilai fungsi pada grid:

- $U_{i,j} = 10$
- $U_{i+1,j} = 12$
- $U_{i-1,j} = 8$
- $U_{i,j+1} = 11$
- $U_{i,j-1} = 9$

Jika $\Delta x = \Delta y = 1$, hitung nilai pendekatan:

$$
\nabla^2 u_{i,j}
$$


## Penyelesaian

### 1. Substitusi ke rumus

$$
\nabla^2 u_{i,j} =
(8 - 2(10) + 12) + (9 - 2(10) + 11)
$$


### 2. Hitung masing-masing bagian

Bagian arah x:
$$
8 - 20 + 12 = 0
$$

Bagian arah y:
$$
9 - 20 + 11 = 0
$$


### 3. Hasil akhir

$$
\nabla^2 u_{i,j} = 0 + 0 = 0
$$


## Penjelasan

- Nilai Laplacian = 0 berarti titik tersebut dalam kondisi seimbang
- Tidak ada perubahan (tidak ada sumber atau sink)


## Intinya

- 5-point stencil menggunakan 5 titik:
  - 1 titik pusat
  - 4 titik tetangga
- Digunakan untuk menghitung turunan kedua secara numerik


## Kesimpulan

Nilai Laplacian di titik $(i,j)$ adalah:

$$
0
$$

yang menunjukkan kondisi stabil (seperti persamaan Laplace).

## Pengurutan Variabel Tak Diketahui

Kita dapat membayangkan masalah dua dimensi kita sebagai terletak pada sebuah grid seperti pada gambar berikut.
Di sini, variabel tak diketahui ditandai dengan warna ungu dan batas-batas ditandai dengan warna hijau :

```python
import numpy as np              # library untuk operasi numerik (array, dll)
import matplotlib.pyplot as plt # library untuk membuat grafik/visualisasi

# =========================
# Parameter Grid
# =========================
m = 12              # jumlah titik interior (titik merah)
n = m + 2           # total titik termasuk boundary (pinggir)

x = np.arange(n)    # array koordinat x (0 sampai n-1)
y = np.arange(n)    # array koordinat y (0 sampai n-1)

# =========================
# Inisialisasi Plot
# =========================
plt.figure(figsize=(6,6))   # membuat figure dengan ukuran 6x6 inch

# =========================
# Plot Titik Grid
# =========================
for i in range(n):          # loop untuk indeks x
    for j in range(n):      # loop untuk indeks y
        
        # cek apakah titik termasuk boundary (pinggir domain)
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            plt.plot(i, j, 'o', color='c')   # plot titik boundary 
        else:
            plt.plot(i, j, 'o', color='m')    # plot titik interior 

# =========================
# Gambar Garis Grid
# =========================
for i in range(n):  # loop setiap garis
    
    plt.plot([0, n-1], [i, i], color='black', linewidth=1)  
    # garis horizontal dari kiri ke kanan
    
    plt.plot([i, i], [0, n-1], color='black', linewidth=1)  
    # garis vertikal dari bawah ke atas

# =========================
# Pengaturan Tampilan
# =========================
plt.gca().set_aspect('equal')   # membuat skala x dan y sama (grid kotak)
plt.xlim(-1, n)                 # batas sumbu x agar ada margin
plt.ylim(-1, n)                 # batas sumbu y agar tidak terpotong

plt.xticks([])                  # menghilangkan angka pada sumbu x
plt.yticks([])                  # menghilangkan angka pada sumbu y

plt.title("2D Grid: Interior (Merah) & Boundary (Biru)")  # judul grafik

# =========================
# Tampilkan Hasil
# =========================
plt.show()   # menampilkan semua plot ke layar
```

# Representasi Grid sebagai Matriks

Kita dapat memandang grid sebagai sebuah matriks:

$$
U=\begin{bmatrix}
U_{1,m}&U_{2,m}&\cdots &U_{m-1,m}&U_{m,m}\\
U_{1,m-1}&U_{2,m-1}&\cdots &U_{m-1,m-1}&U_{m,m-1}\\
\vdots &&&&\vdots \\
U_{1,2}&U_{2,2}&\cdots &U_{m-1,2}&U_{m,2}\\
U_{1,1}&U_{2,1}&\cdots &U_{m-1,1}&U_{m,1}
\end{bmatrix}
$$

## Kenapa tidak sesuai dengan notasi matriks biasa?

Walaupun bentuknya seperti matriks, cara pembacaannya berbeda:

### 1. Matriks biasa
- Elemen $a_{ij}$:
  - $i$ = baris (dari atas ke bawah)
  - $j$ = kolom (dari kiri ke kanan)
- Elemen $a_{11}$ berada di pojok kiri atas


### 2. Grid (numerik/PDE)
- Elemen $U_{i,j}$:
  - $i$ = arah horizontal (x) → kiri ke kanan
  - $j$ = arah vertikal (y) → bawah ke atas
- Elemen $U_{1,1}$ berada di pojok kiri bawah


### 3. Perbedaan utama

| Matriks Biasa | Grid |
|--------------|------|
| Asal di kiri atas | Asal di kiri bawah |
| Indeks ke bawah | Indeks ke atas |
| Struktur array | Representasi ruang (x, y) |


### Kesimpulan

Grid ditulis seperti matriks, tetapi:
- indeksnya mengikuti koordinat fisik (x, y)
- bukan aturan baris–kolom matriks standar

Sehingga penting untuk tidak salah menafsirkan posisi elemen dalam grid.

Ketika kita berada pada satu dimensi, pengurutan variabel tak diketahui (dan persamaan) cukup jelas, tetapi sekarang kita memiliki beberapa pilihan yang harus dibuat.
Mungkin pilihan yang paling jelas adalah memecah array 2D menjadi baris-baris, lalu menambahkan setiap baris ke akhir baris sebelumnya. Hal ini menghasilkan pengurutan:
$$ U=[U_{1,1},U_{2,1},\ldots ,U_{m-1,1},U_{m,1},U_{1,2},U_{2,2},\ldots ,U_{m-1,m},U_{m,m}] $$
Coba lihat apakah kamu dapat menemukan struktur matriks A sekarang dengan pengurutan ini.


## Bentuk Matriks dari Diskretisasi Laplace

Diskretisasi operator Laplace menggunakan 5-point stencil menghasilkan sistem persamaan linear yang dapat dituliskan dalam bentuk matriks:

$$
A = \frac{1}{\Delta x^2}
\begin{bmatrix}
T & I &   &        &        \\
I & T & I &        &        \\
  & I & T & I      &        \\
  &   & \ddots & \ddots & \ddots \\
  &   &        & I & T & I \\
  &   &        &   & I & T
\end{bmatrix}
$$

dengan matriks $T$ didefinisikan sebagai:

$$
T =
\begin{bmatrix}
-4 & 1  &    &        &        \\
1  & -4 & 1  &        &        \\
   & 1  & -4 & 1      &        \\
   &    & \ddots & \ddots & \ddots \\
   &    &        & 1 & -4 & 1 \\
   &    &        &   & 1 & -4
\end{bmatrix}
$$

dan $I$ adalah matriks identitas berukuran $m \times m$.

### Penjelasan

Matriks $A$ merupakan hasil dari penyusunan sistem persamaan linear dari seluruh titik grid pada domain dua dimensi.

- Matriks $T$ merepresentasikan interaksi antar titik dalam satu baris (arah $x$)
- Matriks $I$ merepresentasikan keterkaitan antar baris (arah $y$)
- Struktur matriks $A$ disebut block tridiagonal

Setiap elemen:
- $-4$ berasal dari titik pusat
- $1$ berasal dari kontribusi empat titik tetangga

### Makna Fisik

Bentuk ini menunjukkan bahwa nilai pada suatu titik dipengaruhi oleh empat titik di sekitarnya, sesuai dengan konsep difusi atau penyebaran pada persamaan Laplace.

Kemungkinan lain adalah pengurutan red-black (seperti papan catur).
Di sini kita “memberi warna” pada setiap variabel tak diketahui dalam grid secara bergantian, lalu setiap variabel yang memiliki warna sama dikelompokkan bersama.

```python
import numpy as np              # operasi numerik
import matplotlib.pyplot as plt # visualisasi

# =========================
# Parameter Grid
# =========================
m = 12              # jumlah titik interior
n = m + 2           # total grid termasuk boundary

# =========================
# Inisialisasi Plot
# =========================
plt.figure(figsize=(6,6))   # ukuran figure

# =========================
# Plot Titik
# =========================
for i in range(n):          # loop arah x
    for j in range(n):      # loop arah y
        
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            plt.plot(i, j, 'o', color='black', markersize=6)   # boundary (hitam)
        
        else:
            if (i + j) % 2 == 0:
                plt.plot(i, j, 'o', color='red', markersize=6)     # interior 1
            else:
                plt.plot(i, j, 'o', color='gold', markersize=6)    # interior 2

# =========================
# Garis Grid
# =========================
for i in range(n):
    plt.plot([0, n-1], [i, i], color='gray', linewidth=1)  # garis horizontal
    plt.plot([i, i], [0, n-1], color='gray', linewidth=1)  # garis vertikal

# =========================
# Tampilan
# =========================
plt.gca().set_aspect('equal')  # skala kotak
plt.xlim(-1, n)                # batas x
plt.ylim(-1, n)                # batas y

plt.xticks([])                 # hilangkan label x
plt.yticks([])                 # hilangkan label y

plt.title("Grid 2D")  # judul

plt.show()  # tampilkan plot
```

## Red-Black Ordering

Dengan menggunakan teknik pengurutan red-black, sistem persamaan linear dapat dituliskan dalam bentuk:

$$
\begin{bmatrix}
D & H \\
H^T & D
\end{bmatrix}
\begin{bmatrix}
U_{\mathrm{red}} \\
U_{\mathrm{black}}
\end{bmatrix}
=
\begin{bmatrix}
f_{\mathrm{red}} \\
f_{\mathrm{black}}
\end{bmatrix}
$$

### Penjelasan

- $U_{\mathrm{red}}$ dan $U_{\mathrm{black}}$ adalah vektor solusi yang dipisahkan berdasarkan pola red-black (seperti papan catur)
- $f_{\mathrm{red}}$ dan $f_{\mathrm{black}}$ adalah vektor sumber
- $D$ adalah matriks diagonal berukuran $m^2/2$
- $H$ adalah matriks pita (*banded matrix*) dengan 4 diagonal tak nol
- $H^T$ adalah transpose dari matriks $H$

### Konsep Red-Black

Grid dua dimensi dibagi menjadi dua kelompok:
- Titik merah (red)
- Titik hitam (black)

Seperti pola papan catur, di mana:
- Titik merah hanya berinteraksi dengan titik hitam
- Titik hitam hanya berinteraksi dengan titik merah

Hal ini menyebabkan matriks sistem memiliki struktur blok seperti di atas.

### Keuntungan Pengurutan

Tujuan utama dari pengurutan ini adalah:

- Mengurangi jarak antar elemen tak nol dalam matriks
- Mengurangi fill-in saat proses eliminasi Gauss
- Meningkatkan efisiensi komputasi

### Makna Fill-in

Fill-in adalah munculnya elemen tak nol baru pada posisi yang sebelumnya nol selama proses eliminasi. Semakin sedikit fill-in, semakin efisien penyelesaian sistem.

### Catatan

Dalam praktiknya, pengurutan seperti ini biasanya dilakukan secara otomatis oleh solver numerik modern, sehingga pengguna tidak perlu mengaturnya secara manual.

Selesaikan PDE berikut dengan pengurutan yang kamu pilih:
$$ \nabla ^2u=-2\sin x\sin y\quad \Omega =[0,2\pi ]\times [0,2\pi ]
u(x,0)=0,\quad u(x,2\pi )=0,\quad u(0,y)=0,\quad u(2\pi ,y)=0. $$

```python
import scipy.sparse as sparse        # Mengimpor modul matriks sparse (jarang terisi)
import scipy.sparse.linalg as linalg # Mengimpor solver aljabar linear untuk matriks sparse

# Problem specification
m = 100                               # Jumlah titik grid interior pada arah x dan y
x = numpy.linspace(0, 2.0 * numpy.pi, m + 2)  # Membuat grid x dari 0 sampai 2π (termasuk batas)
y = numpy.linspace(0, 2.0 * numpy.pi, m + 2)  # Membuat grid y dari 0 sampai 2π
delta_x = 2.0 * numpy.pi / (m + 1)             # Jarak antar grid (Δx)

# Construct A
e = numpy.ones(m)                     # Vektor berisi angka 1 sebanyak m elemen
T = sparse.spdiags([e, -4.0 * e, e], [-1, 0, 1], m, m)  
# Matriks tridiagonal untuk operator Laplacian arah x

S = sparse.spdiags([e, e], [-1, 1], m, m)      
# Matriks yang merepresentasikan interaksi grid tetangga atas dan bawah

I = sparse.eye(m)                     # Matriks identitas ukuran m x m

A = sparse.kron(I, T) + sparse.kron(S, I)      
# Membentuk matriks Laplacian 2D menggunakan produk Kronecker

A /= delta_x**2                       # Membagi matriks dengan (Δx²) sesuai diskretisasi PDE

# Right-hand-side
X, Y = numpy.meshgrid(x[1:-1], y[1:-1])  
# Membuat grid 2D tanpa titik batas

f = -2.0 * numpy.sin(X) * numpy.sin(Y)  
# Fungsi sumber f(x,y) dari persamaan Poisson

# Solve
U = numpy.zeros((m+2, m+2))          
# Matriks solusi awal (termasuk batas)

U[1:-1, 1:-1] = linalg.spsolve(A, f.reshape(m**2, order='F')).reshape((m, m), order='F')
# Menyelesaikan sistem linear A*u = f
# reshape digunakan untuk mengubah matriks menjadi vektor dan sebaliknya

# Error
X, Y = numpy.meshgrid(x, y)          
# Membuat grid lengkap termasuk batas

print(numpy.linalg.norm((x[1] - x[0]) * (U - numpy.sin(X) * numpy.sin(Y)), ord=1))
# Menghitung error antara solusi numerik dan solusi eksak
# menggunakan norma L1

# Plot solution
fig = plt.figure()
axes = fig.add_subplot(1, 1, 1)

sol_plot = axes.pcolor(X, Y, U, cmap=plt.get_cmap('RdBu_r'))
# Membuat plot warna (heatmap) dari solusi numerik U

axes.set_title("Solution u(x,y)")    
axes.set_xlabel("x")
axes.set_ylabel("y")

axes.set_xlim((0.0, 2.0 * numpy.pi))
axes.set_ylim((0.0, 2.0 * numpy.pi))
# Mengatur batas sumbu plot

cbar = fig.colorbar(sol_plot, ax=axes)
cbar.set_label("u(x, y)")
# Menambahkan colorbar untuk menunjukkan nilai solusi

fig = plt.figure()
axes = fig.add_subplot(1, 1, 1)

sol_plot = axes.pcolor(X, Y, numpy.abs(U - numpy.sin(X) * numpy.sin(Y)), cmap=plt.get_cmap('RdBu_r'))
# Plot error absolut antara solusi numerik dan solusi eksak

axes.set_title("Error |U - u|")
axes.set_xlabel("x")
axes.set_ylabel("y")

axes.set_xlim((0.0, 2.0 * numpy.pi))
axes.set_ylim((0.0, 2.0 * numpy.pi))

cbar = fig.colorbar(sol_plot, ax=axes)
cbar.set_label("u(x, y)")

plt.show()  
# Menampilkan semua plot
```

## Akurasi dan Stabilitas

Tentu saja kita dapat (dan seharusnya) mengajukan pertanyaan yang sama seperti pada kasus satu dimensi, yaitu apakah skema kita akan konvergen. Untuk melakukan ini kita perlu mempertimbangkan LTE dan stabilitas metode kita.

# Local Truncation Error (LTE)

Kita tahu bahwa LTE didefinisikan sebagai:

$$
\begin{aligned}
\tau_{ij} &= \frac{1}{\Delta x^2} \Big( u(x_{i-1}, y_{j}) + u(x_{i+1}, y_{j}) + u(x_{i}, y_{j-1}) + u(x_{i}, y_{j+1}) - 4 u(x_{i}, y_{j}) \Big) \\
&\quad - f(x_i, y_j)
\end{aligned}
$$

## Penjelasan

Persamaan ini berasal dari pendekatan metode beda hingga (finite difference) untuk persamaan diferensial parsial, seperti persamaan Poisson:

$$
\nabla^2 u = f(x,y)
$$


### 1. Makna Persamaan

Bagian:

$$
\frac{1}{\Delta x^2} \Big( u_{kiri} + u_{kanan} + u_{bawah} + u_{atas} - 4u_{pusat} \Big)
$$

adalah aproksimasi dari operator Laplacian ($\nabla^2 u$) di titik $(i,j)$.

- $u(x_{i-1}, y_j)$ → kiri  
- $u(x_{i+1}, y_j)$ → kanan  
- $u(x_i, y_{j-1})$ → bawah  
- $u(x_i, y_{j+1})$ → atas  
- $u(x_i, y_j)$ → titik pusat  


### 2. Arti $\tau_{ij}$

$\tau_{ij}$ adalah **Local Truncation Error**, yaitu selisih antara:

- hasil pendekatan numerik
- dengan solusi eksak dari persamaan diferensial

Secara konsep:

$$
\tau_{ij} = \text{aproksimasi} - \text{nilai sebenarnya}
$$


### 3. Interpretasi

- $\tau_{ij} = 0$ → tidak ada error (ideal)
- $\tau_{ij}$ kecil → metode akurat
- $\tau_{ij}$ besar → error besar di titik tersebut


### 4. Kesimpulan

- Persamaan ini mengukur kesalahan dari metode beda hingga
- Digunakan untuk mengevaluasi akurasi metode numerik
- Semakin kecil $\Delta x$, biasanya error semakin kecil

# Ekspansi Deret Taylor dan LTE

Untuk menghitung Local Truncation Error (LTE), digunakan ekspansi deret Taylor di sekitar titik $(x_i, y_j)$.


## Arah x

$$
u(x_{i+1},y_j)=u(x_i,y_j)+\Delta x\, u_x+\frac{\Delta x^2}{2}u_{xx}
+\frac{\Delta x^3}{6}u_{xxx}+\frac{\Delta x^4}{24}u_{xxxx}
+\mathcal{O}(\Delta x^5)
$$

$$
u(x_{i-1},y_j)=u(x_i,y_j)-\Delta x\, u_x+\frac{\Delta x^2}{2}u_{xx}
-\frac{\Delta x^3}{6}u_{xxx}+\frac{\Delta x^4}{24}u_{xxxx}
+\mathcal{O}(\Delta x^5)
$$

### Penjelasan arah x

Jika kedua persamaan dijumlahkan:

$$
u(x_{i+1},y_j) + u(x_{i-1},y_j)
$$

maka:
- suku ganjil ($\Delta x$, $\Delta x^3$) saling menghilangkan
- tersisa hanya suku genap

Hasilnya:

$$
u(x_{i+1},y_j) + u(x_{i-1},y_j)
= 2u(x_i,y_j) + \Delta x^2 u_{xx} + \frac{\Delta x^4}{12}u_{xxxx} + \mathcal{O}(\Delta x^6)
$$


## Arah y

$$
u(x_i,y_{j+1})=u(x_i,y_j)+\Delta y\, u_y+\frac{\Delta y^2}{2}u_{yy}
+\frac{\Delta y^3}{6}u_{yyy}+\frac{\Delta y^4}{24}u_{yyyy}
+\mathcal{O}(\Delta y^5)
$$

$$
u(x_i,y_{j-1})=u(x_i,y_j)-\Delta y\, u_y+\frac{\Delta y^2}{2}u_{yy}
-\frac{\Delta y^3}{6}u_{yyy}+\frac{\Delta y^4}{24}u_{yyyy}
+\mathcal{O}(\Delta y^5)
$$

### Penjelasan arah y

Dengan cara yang sama:

$$
u(x_i,y_{j+1}) + u(x_i,y_{j-1})
= 2u(x_i,y_j) + \Delta y^2 u_{yy} + \frac{\Delta y^4}{12}u_{yyyy} + \mathcal{O}(\Delta y^6)
$$


## Menggabungkan (Operator Laplace)

Jika digabungkan arah x dan y:

$$
u_{xx} + u_{yy}
$$

diperoleh bentuk pendekatan:

$$
\frac{u_{i+1,j} + u_{i-1,j} + u_{i,j+1} + u_{i,j-1} - 4u_{i,j}}{\Delta x^2}
$$

(untuk $\Delta x = \Delta y$)


## Kesimpulan

### 1. Suku ganjil hilang
- Karena tanda $+$ dan $-$ saling meniadakan
- Ini membuat aproksimasi lebih stabil

### 2. Suku genap membentuk turunan kedua
- $\Delta x^2 u_{xx}$ dan $\Delta y^2 u_{yy}$
- Inilah asal munculnya operator Laplace

### 3. Error utama

Error yang tersisa berasal dari suku orde lebih tinggi:

$$
\mathcal{O}(\Delta x^2, \Delta y^2)
$$

Artinya:
- metode ini memiliki orde 2 (second order accuracy)
- jika $\Delta x$ dan $\Delta y$ diperkecil → error berkurang secara kuadrat


## Inti yang bisa kamu sampaikan saat presentasi

- Deret Taylor dipakai untuk mendekati nilai di sekitar titik
- Penjumlahan membuat suku ganjil hilang
- Tersisa turunan kedua → membentuk Laplacian
- Error utama berorde 2 → metode cukup akurat

Dengan menggunakan juga deret Taylor pada arah y, kita dapat menuliskan LTE sebagai
$$ \tau _{ij}=\frac{1}{12}\Delta x^2\, (u_{xxxx}+u_{yyyy})+\mathcal{O}(\Delta x^4). $$


# Sistem Linear untuk LTE

Sistem linear untuk error dapat dituliskan sebagai:

$$
A_{\Delta x} E_{\Delta x} = -\tau_{\Delta x}
$$



## Penjelasan

Persamaan ini adalah bentuk sistem linear yang menghubungkan:

- $A_{\Delta x}$ → matriks hasil diskretisasi (finite difference)
- $E_{\Delta x}$ → vektor error (selisih solusi numerik dengan solusi eksak)
- $\tau_{\Delta x}$ → Local Truncation Error (LTE)



## 1. Makna Persamaan

Persamaan ini menyatakan bahwa:

$$
\text{Error global} \quad \propto \quad \text{Error lokal (LTE)}
$$

Artinya:
- LTE ($\tau$) adalah sumber kesalahan
- Error solusi ($E$) adalah akibatnya setelah melalui sistem



## 2. Interpretasi Fisik/Matematis

- $A_{\Delta x}$ bekerja seperti "operator" yang menyebarkan error ke seluruh grid
- Jika $\tau_{\Delta x} = 0$ → maka:
  $$
  A_{\Delta x} E_{\Delta x} = 0 \Rightarrow E_{\Delta x} = 0
  $$
  (artinya solusi numerik = solusi eksak)

- Jika $\tau_{\Delta x} \neq 0$ → akan muncul error pada solusi



## 3. Kenapa ada tanda minus?

Tanda minus muncul karena:
- LTE didefinisikan sebagai selisih antara aproksimasi dan nilai sebenarnya
- Saat dimasukkan ke sistem persamaan, bentuknya menjadi sumber kesalahan dengan tanda negatif



## 4. Tentang Pengurutan Persamaan

Perlu dicatat:

> Pengurutan persamaan tidak berpengaruh terhadap error

Artinya:
- Menukar urutan baris dalam sistem linear (reordering)
- Tidak mengubah nilai solusi $E_{\Delta x}$
- Hanya mengubah cara penulisan matriks, bukan hasil akhirnya


## Kesimpulan

- Persamaan ini menghubungkan LTE dengan error global
- LTE kecil → error solusi juga kecil (metode konvergen)
- Matriks $A$ menentukan bagaimana error menyebar
- Urutan persamaan tidak mempengaruhi hasil error

Untuk stabilitas dalam norma-2 kita sekali lagi dapat mempertimbangkan nilai eigen dari sistem di atas. Nilai eigen tersebut adalah
$$ \lambda _{pq}=\frac{2}{\Delta x^2}\left( (\cos (p\pi \Delta x)-1)+(\cos (q\pi \Delta x)-1)\right) $$
dengan vektor eigen yang bersesuaian
$$ v_{ij}^{p,q}=\sin (p\pi i\Delta x)\sin (q\pi j\Delta x). $$


Karena nilai eigen semuanya bernilai negatif (A pada kenyataannya adalah negative definite), maka nilai eigen yang paling dekat dengan titik asal adalah
$$ \lambda _{11}=-2\pi ^2+\mathcal{O}(\Delta x^2) $$
yang menghasilkan jari-jari spektral

Kita dapat menggunakan batas ini pada $ A^{-1} $ untuk menunjukkan stabilitas dan dengan demikian konvergensi dari diskretisasi.


Sebuah kuantitas serupa dan berguna untuk dipertimbangkan adalah bilangan kondisi dari matriks A. Ingat bahwa ini dapat didefinisikan sebagai
$$ \kappa (A)=\| A\| \, \| A^{-1}\| $$


Dalam norma-2 kita sudah mengetahui beberapa informasi tentang $$ A^{-1}, $$ tetapi kita juga dapat menggunakan ekspresi dari atas untuk menemukan jari-jari spektral dari A. Nilai eigen terbesar di sana adalah
$$ \lambda _{mm}\approx -\frac{8}{\Delta x^2} $$
yang menghasilkan bilangan kondisi
$$ \kappa _2(A)=\frac{4}{\pi ^2\Delta x^2}=\mathcal{O}\left( \frac{1}{\Delta x^2}\right) $$
Maka matriks ini menjadi semakin ill-conditioned ketika $\Delta x\rightarrow 0$.


```python
# ==========================================
# Eigenvalues 2D Laplacian (5-Point Stencil)
# ==========================================

import numpy as np              # Library untuk komputasi numerik (array, matriks, dll)
import matplotlib.pyplot as plt # Library untuk visualisasi/plot

# =========================
# 1. Parameter Grid
# =========================
m = 10                      # jumlah titik grid internal (tanpa boundary)
L = 1.0                     # panjang domain (misalnya dari 0 sampai 1)
dx = L / (m + 1)            # jarak antar grid (discretization step)

# =========================
# 2. Hitung Eigenvalues
# =========================
# Membuat array kosong untuk menyimpan semua eigenvalue
eig_vals = np.zeros(m * m)

# Loop untuk semua mode (p, q)
for i, p in enumerate(range(1, m + 1)):      # mode arah x
    for j, q in enumerate(range(1, m + 1)):  # mode arah y
        
        # Rumus eigenvalue Laplacian 2D (finite difference 5-point stencil)
        eig_vals[i * m + j] = (2.0 / dx**2) * (
            (np.cos(p * np.pi * dx) - 1.0) +
            (np.cos(q * np.pi * dx) - 1.0)
        )

# =========================
# 3. Informasi Statistik
# =========================
# Menampilkan jumlah total eigenvalue (harusnya m^2)
print("Jumlah eigenvalue :", len(eig_vals))

# Menampilkan nilai eigenvalue minimum
print("Minimum eigenvalue:", np.min(eig_vals))

# Menampilkan nilai eigenvalue maksimum
print("Maximum eigenvalue:", np.max(eig_vals))

# =========================
# 4. Plot Eigenvalues (1D)
# =========================
plt.figure()

# Plot eigenvalue di sumbu real (bagian imajiner ≈ 0)
plt.plot(eig_vals, np.zeros_like(eig_vals), 'o')

plt.title("Eigenvalues of 2D Laplacian (5-point stencil)")
plt.xlabel("Eigenvalue")                     # sumbu x = nilai eigenvalue
plt.ylabel("Imaginary Part (≈ 0)")           # sumbu y (nol karena real)
plt.grid(True)

# =========================
# 5. Plot Histogram
# =========================
plt.figure()

# Histogram untuk melihat distribusi eigenvalue
plt.hist(eig_vals, bins=20)

plt.title("Distribution of Eigenvalues")
plt.xlabel("Eigenvalue")
plt.ylabel("Frequency")                      # frekuensi kemunculan
plt.grid(True)

# =========================
# 6. Visualisasi 2D (Matrix Mode)
# =========================
# Mengubah eigenvalue menjadi matriks 2D (m x m)
eig_matrix = eig_vals.reshape((m, m))

plt.figure()

# Visualisasi dalam bentuk heatmap
plt.imshow(eig_matrix)

plt.colorbar()                               # menampilkan skala warna
plt.title("Eigenvalue Distribution (2D Grid)")
plt.xlabel("q index")                         # indeks arah y
plt.ylabel("p index")                         # indeks arah x

# =========================
# 7. Tampilkan Semua Plot
# =========================
plt.show()                                    # menampilkan semua grafik
```

## 9-Point Stencil

Pendekatan lain untuk mendiskritisasi operator Laplace $\nabla^2$ adalah menggunakan 9-point stencil, yang melibatkan titik pusat, titik tetangga langsung, dan titik diagonal.

$$
\nabla^2 u_{ij} =
\frac{1}{6\Delta x^2}
\left(
4u_{i-1,j} + 4u_{i+1,j} + 4u_{i,j-1} + 4u_{i,j+1}
+ u_{i-1,j-1} + u_{i+1,j-1} + u_{i+1,j+1} + u_{i-1,j+1}
- 20u_{i,j}
\right)
$$

### Penjelasan

Persamaan ini merupakan pengembangan dari 5-point stencil, dengan tambahan kontribusi dari titik diagonal.

#### Titik yang digunakan:
- 1 titik pusat: $u_{i,j}$
- 4 titik tetangga langsung:  
  $u_{i\pm1,j}$ dan $u_{i,j\pm1}$
- 4 titik diagonal:  
  $u_{i\pm1,j\pm1}$

Total = 9 titik → disebut *9-point stencil*


### Arti Koefisien

- Koefisien 4 → untuk tetangga langsung (arah x dan y)
- Koefisien 1 → untuk titik diagonal
- Koefisien -20 → untuk titik pusat

Koefisien ini diperoleh dari pengembangan deret Taylor agar menghasilkan aproksimasi Laplacian yang lebih akurat.


### Keunggulan

Dibandingkan 5-point stencil:

- Lebih akurat
- Error lebih kecil
- Memiliki orde akurasi lebih tinggi (mendekati $\mathcal{O}(\Delta x^4)$ dalam kondisi tertentu)


### Intuisi Fisik

Metode ini mempertimbangkan pengaruh:
- arah horizontal & vertikal
- serta arah diagonal

Sehingga distribusi nilai menjadi lebih “halus” dan mendekati solusi kontinu.


### Kesimpulan

9-point stencil adalah metode diskretisasi Laplace yang:
- menggunakan lebih banyak titik (9 titik)
- memberikan hasil lebih akurat
- cocok untuk simulasi numerik dengan presisi tinggi

## Local Truncation Error (LTE) – 9-Point Stencil

Local Truncation Error (LTE) untuk metode 9-point stencil diberikan oleh:

$$
\tau_{ij} =
\frac{\Delta x^2}{12}
\left(
u_{xxxx} + 2u_{xxyy} + u_{yyyy}
\right)
+ \mathcal{O}(\Delta x^4)
$$

### Penjelasan

Persamaan ini menunjukkan galat lokal yang muncul akibat aproksimasi operator Laplace menggunakan metode 9-point stencil.

- $\tau_{ij}$ = error lokal pada titik grid $(i,j)$  
- $\Delta x$ = jarak antar grid  
- $u_{xxxx}, u_{yyyy}$ = turunan orde empat  
- $u_{xxyy}$ = turunan campuran orde empat  


### Makna Matematis

Suku:
$$
u_{xxxx} + 2u_{xxyy} + u_{yyyy}
$$

sebenarnya dapat ditulis sebagai:

$$
\nabla^4 u
$$

yang disebut operator biharmonik.


### Intuisi

- Error tidak lagi berasal dari turunan orde dua (seperti 5-point stencil)
- Tetapi dari turunan orde empat
- Artinya pendekatan ini jauh lebih mendekati solusi asli


### Orde Akurasi

- LTE utama berorde:
$$
\mathcal{O}(\Delta x^2)
$$

- Tetapi karena struktur stencil lebih baik, metode ini sering menghasilkan solusi yang lebih akurat dibanding 5-point stencil


### Perbandingan dengan 5-Point Stencil

| Metode          | Titik | Error Utama | Akurasi |
|-----------------|------|------------|--------|
| 5-point stencil | 5    | orde 2     | standar |
| 9-point stencil | 9    | orde lebih tinggi | lebih akurat |


### Kesimpulan

Metode 9-point stencil memiliki LTE yang bergantung pada turunan orde empat, sehingga memberikan hasil aproksimasi yang lebih halus dan akurat dibanding metode 5-point stencil.

Jadi apa yang kita peroleh? Jika kita melihat pada $$\tau _{ij}$$  
kita sebenarnya dapat menuliskan suku turunannya sebagai
$$ u_{xxxx}+2u_{xxyy}+u_{yyyy}=\nabla ^2(\nabla ^2u)\equiv \nabla ^4u, $$
yaitu Laplasian dari Laplasian $$ (disebut operator biharmonik)$$  
Jika kita kembali ke PDE asli, kita dapat langsung menemukan bentuk galat ini dengan mencatat bahwa
$$ \begin{aligned}\nabla ^2u&=f\\ \nabla ^2(\nabla ^2u)&=\nabla ^2f\end{aligned} $$
Dengan demikian, kita dapat menghitung suku galat secara langsung dengan mengambil Laplasian dari ruas kanan PDE asli!


Ini juga dapat menghasilkan metode orde lebih tinggi. Jika kita mendefinisikan
$$ f_{ij}=f(x_i,y_j)+\frac{\Delta x^2}{12}\nabla ^2f(x_i,y_j) $$
maka kita dapat menghilangkan suku berikutnya dari $$ Local Truncation Error (LTE)!$$ Kita bahkan bisa melakukan ini meskipun hanya memiliki fungsi $ f(x,y) $ pada sekumpulan titik diskret, karena kita dapat menggunakan $ 5-point stencil.$
